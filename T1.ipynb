```python
!pip install openpyxl

from google.colab import files
import pandas as pd
import numpy as np
import tensorflow as tf
from sklearn.preprocessing import MinMaxScaler

# Subir el archivo Excel
subiditos = files.upload()

# Obtener nombre del archivo
archivito = next(iter(subiditos))

# Mostrar hojas disponibles
excelito = pd.ExcelFile(archivito)
print("\nHojas disponibles:", excelito.sheet_names)

# Seleccionar hoja
hojita_seleccionada = input("\nIngresa el nombre de la hoja a usar: ")

# Leer datos
dataframecito = pd.read_excel(archivito, sheet_name=hojita_seleccionada)

# 1. Eliminar valores nulos
dataframecito_limpio = dataframecito.dropna()

# 2. Convertir columna de género
columnita_genero = dataframecito_limpio.columns[1]
dataframecito_limpio[columnita_genero] = dataframecito_limpio[columnita_genero].replace({
    'FEMENINO': 0,
    'MASCULINO': 1
})

# 3. Normalización entre 0-1
columnitas_normalizar = dataframecito_limpio.columns.difference([columnita_genero])
escaloncito = MinMaxScaler()
dataframecito_limpio[columnitas_normalizar] = escaloncito.fit_transform(dataframecito_limpio[columnitas_normalizar])

# 4. Convertir a numpy array y tensor
arreglito_numpy = dataframecito_limpio.values.astype(np.float32)
tensorcito = tf.convert_to_tensor(arreglito_numpy, dtype=tf.float32)

# Mostrar resultados
print("\nDatos como Numpy Array:")
print(arreglito_numpy)
print("\nForma del array:", arreglito_numpy.shape)
print("\nDatos como Tensor:")
print(tensorcito)
```
